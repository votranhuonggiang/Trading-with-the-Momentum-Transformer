```python
from pathlib import Path
import subprocess, sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent
CFG = PROJECT_ROOT / 'configs' / 'default.yaml'
print(PROJECT_ROOT)
```

```python
cmd = [sys.executable, str(PROJECT_ROOT / 'src' / 'feature_engineering.py'), '--config', str(CFG)]
print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)
print('Done feature engineering.')
```

```python
feat_path = PROJECT_ROOT / 'data' / 'features' / 'features_5min.parquet'
df = pd.read_parquet(feat_path)
print('shape:', df.shape)
print(df.head())
```

```python
null_ratio = df.isnull().mean().sort_values(ascending=False)
print('Top null ratio columns:')
print(null_ratio.head(20))
```

```python
target_cols = [c for c in df.columns if 'target' in c or 'future_return_sign' in c]
print('target columns:', target_cols)
print(df[target_cols].describe(include='all'))
```
